## Paso 2 — Python como Calculadora de Ingeniería y `print()`
En Excel escribes `=A1*B1` dentro de una celda. En Python puedes hacer cálculos exactos, documentarlos con comentarios (`#`) para que tus colegas los entiendan y mostrar resultados con `print()`.

```python
# 2. Cálculos operativos de una bomba centrífuga y costo eléctrico diario
# Nota: cualquier línea que empieza con '#' es un comentario humano (Python la ignora)

caudal_m3_h = 120.0          # Caudal volumétrico en metros cúbicos por hora
potencia_kw = 45.5           # Potencia consumida en el eje del motor (kW)
horas_turno = 24             # Operación continua 24 horas al día
tarifa_kwh_usd = 0.11        # Tarifa eléctrica industrial en USD por kWh

# Conversiones y KPIs diarios
caudal_litros_min = (caudal_m3_h * 1000) / 60
energia_diaria_kwh = potencia_kw * horas_turno
costo_diario_usd = energia_diaria_kwh * tarifa_kwh_usd
costo_anual_usd = costo_diario_usd * 365

print("Caudal equivalente (L/min):", round(caudal_litros_min, 1))
print("Consumo diario (kWh):", energia_diaria_kwh)
print("Costo eléctrico diario (USD):", round(costo_diario_usd, 2))
print("Costo eléctrico anual estimado (USD):", round(costo_anual_usd, 2))
```
