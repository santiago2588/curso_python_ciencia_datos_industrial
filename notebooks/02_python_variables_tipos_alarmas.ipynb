## Paso 2 — Conversión de Unidades de Ingeniería y `f-strings`
En plantas industriales mezclamos manuales OEM americanos (°F, psi, HP) con estándares internacionales (°C, bar, kW). Con `f-strings` (`f"... {valor:.2f} ..."`) imprimimos reportes limpios y profesionales.

```python
# 2. Conversión desde hoja de datos OEM imperial hacia Sistema Internacional (SI)
temp_entrada_f = 365.0
presion_diseno_psi = 175.0
potencia_motor_hp = 250.0

# Fórmulas físicas exactas
temp_entrada_c = (temp_entrada_f - 32.0) * (5.0 / 9.0)
presion_diseno_bar = presion_diseno_psi * 0.0689476
potencia_motor_kw = potencia_motor_hp * 0.7457

print("=== FICHA TÉCNICA NORMALIZADA (SI) ===")
print(f"Temperatura proceso : {temp_entrada_f:.1f} °F  = {temp_entrada_c:.2f} °C")
print(f"Presión de diseño   : {presion_diseno_psi:.1f} psi = {presion_diseno_bar:.2f} bar")
print(f"Potencia eléctrica  : {potencia_motor_hp:.1f} HP  = {potencia_motor_kw:.2f} kW")
```

## Paso 3 — Operadores Lógicos (`and`, `or`, `not`) para Permisivos de Arranque
Un enclavamiento de seguridad (*interlock*) verifica múltiples condiciones simultáneas antes de autorizar el arranque de una bomba o turbina.

```python
# 3. Evaluación de permisivo de arranque seguro para bomba centrífuga BOM-02
nivel_tanque_pct = 68.0
presion_succion_bar = 2.4
paro_emergencia_activo = False

# Reglas: nivel >= 30% Y presión succión >= 1.5 bar Y NO hay paro de emergencia
nivel_ok = nivel_tanque_pct >= 30.0
presion_ok = presion_succion_bar >= 1.5
seguridad_ok = not paro_emergencia_activo

permisivo_arranque = nivel_ok and presion_ok and seguridad_ok

print(f"Nivel suficiente (>=30%)  : {nivel_ok}")
print(f"Presión succión (>=1.5b)  : {presion_ok}")
print(f"Sin paro de emergencia    : {seguridad_ok}")
print(f"¿AUTORIZAR ARRANQUE?      : {permisivo_arranque}")
```

## Paso 4 — Semáforo Multinivel con `if / elif / else`
Observa la **indentación** (4 espacios a la derecha): en Python, esa sangría indica qué instrucciones pertenecen dentro de cada condición.

```python
# 4. Clasificador de estado operativo por vibración (ISO 10816) y temperatura
vibracion_mms = 4.6
temperatura_cojinete_c = 82.0

if vibracion_mms >= 7.1 or temperatura_cojinete_c >= 95.0:
    estado = "DISPARO CRÍTICO"
    accion = "Detener equipo inmediatamente y generar OT de emergencia."
elif vibracion_mms >= 4.5 or temperatura_cojinete_c >= 85.0:
    estado = "ADVERTENCIA ALTA"
    accion = "Inspeccionar alineación y lubricación en las próximas 24 horas."
elif vibracion_mms >= 2.8:
    estado = "PRECAUCIÓN"
    accion = "Mantener monitoreo de tendencia en bitácora de turno."
else:
    estado = "NORMAL"
    accion = "Operación dentro de envolvente nominal."

print(f"Lectura -> Vibración: {vibracion_mms} mm/s | Temp cojinete: {temperatura_cojinete_c} °C")
print(f"Estado  -> {estado}")
print(f"Acción  -> {accion}")
```

## Paso 5 — Error Lab 02: `TypeError` y `ZeroDivisionError`
Cuando importas un reporte CSV sucio, un número puede llegar como texto (`"185.5"`), provocando un **`TypeError`** al sumar. Y cuando una bomba está detenida (`flujo = 0.0`), dividir potencia entre flujo dispara un **`ZeroDivisionError`**.

```python
# 5. Error Lab 02: detectando y resolviendo TypeError y ZeroDivisionError
lectura_csv = "185.5"   # Llegó como texto (str) en lugar de decimal (float)

print("--- Demo 1: TypeError (texto + número) ---")
try:
    resultado_mal = lectura_csv + 5.0
except TypeError as err:
    print("Error capturado -> TypeError:", err)
    lectura_num = float(lectura_csv)
    print(f"Corregido con float() -> {lectura_num + 5.0:.1f} °C")

print("\n--- Demo 2: ZeroDivisionError (equipo detenido) ---")
potencia_bomba_kw = 18.0
flujo_medido_m3h = 0.0
try:
    consumo_especifico = potencia_bomba_kw / flujo_medido_m3h
except ZeroDivisionError as err:
    print("Error capturado -> ZeroDivisionError:", err)
    if flujo_medido_m3h > 0:
        consumo_especifico = potencia_bomba_kw / flujo_medido_m3h
    else:
        consumo_especifico = 0.0
    print(f"Corregido con guarda 'if flujo > 0' -> {consumo_especifico:.2f} kWh/m3")
```

2. **Reto 2.2:** Convierte `25.0 bar` a `psi` (`bar * 14.5038`) e imprime el resultado con 1 decimal usando `f-string`.

```python
# Soluciones verificadas — Tu Turno Módulo 2
# Reto 2.1
ph_medido = 5.8
if ph_medido < 6.5:
    diag_ph = "ÁCIDO (Dosificar NaOH)"
elif ph_medido <= 8.5:
    diag_ph = "CONFORME"
else:
    diag_ph = "ALCALINO (Dosificar H2SO4)"
print(f"Reto 2.1 -> pH = {ph_medido:.1f} -> {diag_ph}")

# Reto 2.2
presion_bar = 25.0
presion_psi = presion_bar * 14.5038
print(f"Reto 2.2 -> {presion_bar:.1f} bar equivalen a {presion_psi:.1f} psi")
```
